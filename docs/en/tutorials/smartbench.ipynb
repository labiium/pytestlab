```python
from contextlib import ExitStack

from pytestlab import AutoInstrument

# Independent simulation profiles; no connected instruments are required.
resources = ExitStack()
awg = AutoInstrument.from_preset("keysight/EDU33212A", simulate=True)
resources.callback(awg.close)
psu = AutoInstrument.from_preset("keysight/EDU36311A", simulate=True)
resources.callback(psu.close)
dmm = AutoInstrument.from_preset("keysight/EDU34450A", simulate=True)
resources.callback(dmm.close)
```

```python
print(awg.id())
print(psu.id())
print(dmm.id())
```

```python
resources.close()
```
